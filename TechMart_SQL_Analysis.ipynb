## SQL Analysis

The cleaned TechMart data was analyzed to answer the key business questions defined for the project.

### 1. Employee Performance by Location

This analysis compares employee sales performance across store locations to identify locations with stronger overall employee performance.

```python
query = """
SELECT
    store_location,
    COUNT(employee_id) AS employee_count,
    SUM(sales_performance) AS total_sales_performance,
    AVG(sales_performance) AS average_sales_performance
FROM Employee_Records
GROUP BY store_location
ORDER BY total_sales_performance DESC;
"""

df = pd.read_sql_query(query, conn)
display(df)
```

**Key Insight:**

Phoenix has the highest total employee sales performance at 53,000, followed by New York at 48,000 and London at 45,000.

However, Chicago has the highest average sales performance per employee at 3,875, despite having only 8 employees. This suggests that Chicago has fewer employees but stronger average individual performance.

### Employee Performance Ranking

Employees were ranked within each store location based on their sales performance.

```python
query = """
SELECT
    employee_id,
    role,
    store_location,
    sales_performance,
    RANK() OVER (
        PARTITION BY store_location
        ORDER BY sales_performance DESC
    ) AS performance_rank
FROM Employee_Records
WHERE sales_performance IS NOT NULL
ORDER BY store_location, performance_rank;
"""

df = pd.read_sql_query(query, conn)
display(df)
```

**Employee Performance Insight:**

Employee performance varies within each store location. The ranking analysis identifies the highest-performing employees within each location while accounting for ties in sales performance.

### 2. Top-Selling Products by Category

This analysis identifies the top-selling products within each product category based on total quantity sold.

```python
query = """
WITH product_sales AS (
    SELECT
        p.category,
        p.product_id,
        p.product_name,
        SUM(s.quantity) AS total_quantity_sold
    FROM Product_Details AS p
    JOIN Sales_Transactions AS s
        ON p.product_id = s.product_id
    WHERE s.quantity IS NOT NULL
    GROUP BY
        p.category,
        p.product_id,
        p.product_name
),
ranked_products AS (
    SELECT
        category,
        product_id,
        product_name,
        total_quantity_sold,
        RANK() OVER (
            PARTITION BY category
            ORDER BY total_quantity_sold DESC
        ) AS category_rank
    FROM product_sales
)
SELECT
    category,
    product_id,
    product_name,
    total_quantity_sold,
    category_rank
FROM ranked_products
WHERE category_rank = 1
ORDER BY category;
"""

df = pd.read_sql_query(query, conn)
display(df)
```

**Key Insight:**

The Keyboard is the top-selling product in the Accessories category, with 111 units sold.

In the Electronics category, the Mouse and Tablet are tied as the top-selling products, with 74 units sold each. This indicates that both products have similar sales demand and should be considered when planning Electronics inventory.

### 3. Customer Purchasing Behavior

This analysis examines customer purchasing frequency and total spending to identify high-value and frequent customers.

```python
query = """
SELECT
    c.customer_id,
    c.location,
    c.loyalty_program,
    COUNT(s.transaction_id) AS total_transactions,
    SUM(s.total_amount) AS total_spent
FROM Customer_Demographics AS c
JOIN Sales_Transactions AS s
    ON c.customer_id = s.customer_id
WHERE s.total_amount IS NOT NULL
GROUP BY
    c.customer_id,
    c.location,
    c.loyalty_program
ORDER BY total_spent DESC;
"""

df = pd.read_sql_query(query, conn)
display(df)
```

**Key Insight:**

Customer 13 is the highest-spending customer in the dataset, with total spending of 440 across 8 transactions. Customer 12 follows with 400 across 6 transactions and is enrolled in the loyalty program.

Several high-spending customers are located in San Antonio, suggesting that this location may warrant further investigation. However, customers with missing loyalty-program information should not be classified as either members or non-members.

### 4. Electronics and Accessories Sales Analysis

This analysis focuses on Electronics and Accessories transactions by employee, store location, and customer loyalty status. The analysis compares transaction volume and revenue to identify sales patterns across these dimensions.

```python
query = """
SELECT
    e.store_location,
    s.employee_id,
    c.loyalty_program,
    p.category,
    COUNT(s.transaction_id) AS total_transactions,
    SUM(s.quantity) AS total_quantity_sold,
    SUM(s.total_amount) AS total_revenue
FROM Sales_Transactions AS s
JOIN Product_Details AS p
    ON s.product_id = p.product_id
JOIN Customer_Demographics AS c
    ON s.customer_id = c.customer_id
JOIN Employee_Records AS e
    ON s.employee_id = e.employee_id
WHERE p.category IN ('Electronics', 'Accessories')
  AND s.quantity IS NOT NULL
  AND s.total_amount IS NOT NULL
GROUP BY
    e.store_location,
    s.employee_id,
    c.loyalty_program,
    p.category
ORDER BY total_revenue DESC;
"""

df = pd.read_sql_query(query, conn)
display(df)
```

**Key Insight:**

Houston employee 45 generated the highest revenue among the displayed Electronics and Accessories combinations, generating 1,060 in Electronics revenue from 14 transactions and 880 in Accessories revenue from 12 transactions.

Los Angeles also shows strong Electronics performance, with employee 2 generating 970 in revenue from 19 transactions and employee 15 generating 850 from loyalty-program customers.

The results demonstrate differences in revenue across employees, store locations, product categories, and customer loyalty status. Missing loyalty-program values should be treated separately and not assumed to represent non-members.

### 5. Store-Category Revenue Ranking

This analysis calculates total revenue for each store and product-category combination and ranks the combinations from highest to lowest revenue.

```python
query = """
WITH store_category_sales AS (
    SELECT
        e.store_location,
        p.category,
        SUM(s.total_amount) AS total_revenue
    FROM Sales_Transactions AS s
    JOIN Product_Details AS p
        ON s.product_id = p.product_id
    JOIN Employee_Records AS e
        ON s.employee_id = e.employee_id
    WHERE p.category IN ('Electronics', 'Accessories')
      AND s.total_amount IS NOT NULL
    GROUP BY
        e.store_location,
        p.category
)
SELECT
    store_location,
    category,
    total_revenue,
    RANK() OVER (
        ORDER BY total_revenue DESC
    ) AS revenue_rank
FROM store_category_sales
ORDER BY revenue_rank;
"""

df = pd.read_sql_query(query, conn)
display(df)
```

**Key Insight:**

Los Angeles generated the highest revenue across both analyzed categories, with Electronics generating 5,510 and Accessories generating 4,190. Electronics was the highest-revenue category in every store, while London had the lowest revenue in both Electronics and Accessories.

The results indicate that Los Angeles is the strongest store location for Electronics and Accessories sales, while London may require further investigation to understand its comparatively lower revenue performance.
